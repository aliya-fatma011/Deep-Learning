# Backpropagation Through Time (BPTT) in RNN
This notebook explains forward propagation, loss calculation, backpropagation through time, and a TensorFlow example.

```
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

X = np.array([
    [[1],[2],[3]],
    [[2],[3],[4]],
    [[3],[4],[5]],
    [[4],[5],[6]]
], dtype=np.float32)

y = np.array([4,5,6,7], dtype=np.float32)

model = Sequential([
    SimpleRNN(8, activation='tanh', input_shape=(3,1)),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')
model.fit(X, y, epochs=20, verbose=1)

test = np.array([[[5],[6],[7]]], dtype=np.float32)
print("Prediction:", model.predict(test))
```


## BPTT Flow

1. Forward pass:
   - x1 → h1
   - x2 + h1 → h2
   - x3 + h2 → h3
   - h3 → output

2. Compute Loss:
   - MSE = (Actual − Predicted)^2

3. Backward pass:
   - Error flows from h3 → h2 → h1.
   - Gradients are computed for Wxh, Whh, and Why.
   - Weights are updated using Gradient Descent.


```
with tf.GradientTape() as tape:
    pred = model(X)
    loss = tf.reduce_mean(tf.square(tf.reshape(y, (-1,1)) - pred))

grads = tape.gradient(loss, model.trainable_variables)

for var, grad in zip(model.trainable_variables, grads):
    print(var.name)
    print(grad)
```
