
# RNN Forward Propagation (Hinglish)

## Formula
h_t = tanh(Wxh*x_t + Whh*h_(t-1) + bh)
y_t = Why*h_t + by

### Step 1
Input sequence: x1, x2, x3

### Step 2
Initialize hidden state:
h0 = 0

### Step 3
For x1:
h1 = tanh(Wxh*x1 + Whh*h0 + bh)
y1 = Why*h1 + by

### Step 4
For x2:
h2 = tanh(Wxh*x2 + Whh*h1 + bh)
y2 = Why*h2 + by

### Step 5
For x3:
h3 = tanh(Wxh*x3 + Whh*h2 + bh)
y3 = Why*h3 + by

## Explanation (Hinglish)

- x_t = Current input
- h_(t-1) = Previous memory
- Wxh = Input to Hidden weights
- Whh = Hidden to Hidden weights
- Why = Hidden to Output weights
- bh, by = Bias
- tanh = Activation function (-1 to +1)

RNN har naye input ke saath previous hidden state ko use karta hai,
isliye usse previous context yaad rehta hai.
