#7.Morphological Operations

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

q = '/kaggle/input/cat-and-dog/training_set/training_set/cats/cat.10.jpg'

# Step 1: Read image
image = cv2.imread(q)

# Step 2: Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Step 3: Threshold
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Step 4: Kernel
kernel = np.ones((3,3), np.uint8)

# Step 5: Morphological operations
erosion = cv2.erode(binary, kernel, iterations=1)
dilation = cv2.dilate(binary, kernel, iterations=1)
opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

# Step 6: Show results
titles = ['Binary','Erosion','Dilation','Opening','Closing']
images = [binary, erosion, dilation, opening, closing]

plt.figure(figsize=(10,6))
for i in range(5):
    plt.subplot(2,3,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.show()
```

#🌫️ 9. Adding Noise

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

q = '/kaggle/input/cat-and-dog/training_set/training_set/cats/cat.10.jpg'

image = cv2.imread(q)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Generate Gaussian noise
mean = 0
sigma = 25
gaussian = np.random.normal(mean, sigma, image.shape)

noisy_img = image + gaussian
noisy_img = np.clip(noisy_img, 0, 255).astype(np.uint8)

plt.imshow(noisy_img)
plt.title("Gaussian Noise")
plt.axis('off')
plt.show()
```

```python
def add_salt_pepper(image, prob=0.02):
    noisy = image.copy()

    # Salt
    salt = np.random.rand(*image.shape[:2]) < prob
    noisy[salt] = 255

    # Pepper
    pepper = np.random.rand(*image.shape[:2]) < prob
    noisy[pepper] = 0

    return noisy

sp_img = add_salt_pepper(image)

plt.imshow(sp_img)
plt.title("Salt & Pepper Noise")
plt.axis('off')
plt.show()
```

```python
speckle = image + image * np.random.randn(*image.shape)

speckle = np.clip(speckle, 0, 255).astype(np.uint8)

plt.imshow(speckle)
plt.title("Speckle Noise")
plt.axis('off')
plt.show()
```

#🔀 10. Shearing (Tilting Image)

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

q = '/kaggle/input/cat-and-dog/training_set/training_set/cats/cat.10.jpg'

# Step 1: Read image
image = cv2.imread(q)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

rows, cols, ch = image.shape

# Step 2: Define shear matrix
shear_factor = 0.3

M = np.float32([
    [1, shear_factor, 0],
    [0, 1, 0]
])

# Step 3: Apply transformation
sheared = cv2.warpAffine(image, M, (cols, rows))

# Step 4: Show result
plt.imshow(sheared)
plt.title("Sheared Image")
plt.axis('off')
plt.show()
```

First, let's define a simple Convolutional Neural Network (CNN) model. This model will consist of convolutional layers, pooling layers, and dense layers for classification.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(512, activation='relu'),
    Dense(1, activation='sigmoid') # Output layer for binary classification (cat or dog)
])
```

Next, we need to compile the model. We'll use the Adam optimizer, binary cross-entropy loss (suitable for binary classification), and accuracy as a metric.

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
```
